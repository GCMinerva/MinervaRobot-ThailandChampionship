```python
import os
import shutil
from pathlib import Path

base_path = '/kaggle/working/dataset'
os.makedirs(f'{base_path}/images/train', exist_ok=True)
os.makedirs(f'{base_path}/labels/train', exist_ok=True)

img_src = '/kaggle/input/ftc2025-2026-decode-ball-dataset/Mixdataset/Mixdataset'
lbl_src = '/kaggle/input/ftc2025-2026-decode-ball-dataset/label/label'

for img_file in Path(img_src).glob('*.jpg'):
    lbl_file = Path(lbl_src) / f'{img_file.stem}.txt'
    if lbl_file.exists():
        shutil.copy(img_file, f'{base_path}/images/train/')
        shutil.copy(lbl_file, f'{base_path}/labels/train/')

data_yaml = {
    'path': '/kaggle/working/dataset',
    'train': 'images/train',
    'val': 'images/train',
    'names': {0: 'purpleball', 1: 'greenball'}
}

import yaml
with open('/kaggle/working/data.yaml', 'w') as f:
    yaml.dump(data_yaml, f)

from ultralytics import YOLO
model = YOLO('yolov8n.pt')
model.train(data='/kaggle/working/data.yaml', epochs=20, imgsz=640, device=[0,1], batch=32)

model.export(format='tflite', imgsz=640, int8=True, data='/kaggle/working/data.yaml')
```
